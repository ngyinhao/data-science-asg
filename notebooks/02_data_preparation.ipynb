## Decisions

- Date was converted into a true date type and expanded into month, day, weekday, and weekend fields.
- Seasons, Holiday, and Functioning Day are encoded through the modelling pipeline with one-hot encoding.
- Non-functioning-day records are kept because a closure or inactive system state is important for bike supply planning.
- Outliers are kept because peak demand is exactly what the supply planning frame needs to understand.
- Numeric features are scaled so Multiple Linear Regression has a fair baseline setup.
- The project uses an 80/20 train-test split with a fixed random seed for reproducibility.
